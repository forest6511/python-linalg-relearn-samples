## LESSON 04 2本の矢印を混ぜると、どこまで届くか

本書の「LESSON 04 2本の矢印を混ぜると、どこまで届くか」を読みながら、下のコードを上から順に1つずつ ▶ で試してください。

### 検証

```python
v = np.array([3, 2])
w = np.array([1, 4])
print(2 * v + 1 * w)
```

```python
rng = np.random.default_rng(42)
v = np.array([3, 2])
w = np.array([1, 4])

c = rng.uniform(-1, 1, size=(100, 2))
reach = c[:, 0:1] * v + c[:, 1:2] * w

print(reach.shape)
print(np.round(reach[:3], 2))
```

```python
plt.scatter(reach[:, 0], reach[:, 1], s=26,
            color="tab:gray", alpha=0.75)
plt.quiver(0, 0, v[0], v[1], angles="xy", scale_units="xy",
           scale=1, color="tab:blue")
plt.quiver(0, 0, w[0], w[1], angles="xy", scale_units="xy",
           scale=1, color="tab:orange")
plt.xlim(-10, 10)
plt.ylim(-8, 8)
plt.gca().set_aspect("equal")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.title("向きが違う2本")
plt.show()
```

### ちょっと変えてみる

**ここからが本番です。** 本文の指示どおりに数字を1か所だけ書きかえて ▶ を押し、結果がどう変わるかを確かめてください。

```python
rng = np.random.default_rng(42)
v = np.array([3, 2])
w = np.array([6, 4])

c = rng.uniform(-1, 1, size=(100, 2))
reach = c[:, 0:1] * v + c[:, 1:2] * w

print(np.round(reach[:3], 2))
```
